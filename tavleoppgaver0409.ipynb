##### Tavleoppgaver
a) Lag en liste som inneholder fargene: "rød", "grønn", "blå". Bruk en for funksjon for å printe ut alle farger bortsett fra "grønn".

b) Her er en liste over alderen til personer som prøvde å kjøre berg og dalbane:

100, 120, 125, 155, 160.

Filtrer ut de under 115cm som var for lave og 125cm for å kjøre alene.

c) Lag en løkke som printer ut alle tallene fra 0 til 10. Når den har gjort det print ut "suksess, avslutter programmet"

```python
#tavleoppgave A

list = ["rød", "grønn", "blå"]

for i in list:
    if i == "grønn":
        continue
    print(i)
```

```python
#tavleoppgave B

list = [100, 120, 125, 155, 160]

for i in list:
    if i < 115:
        print(i, "du er desverre for liten og kan ikke kjøre")
        continue
    elif i < 125:
        print(i, "du må desverre ha med en voksen for å kjøre")
        continue
    else:
        print(i, "du er stor nok til å kjøre alene")
        continue
```

```python
#tavleoppgave C

x = 0

while x < 11:
    print(x)
    x = x + 1
print("suksess, avslutter programmet")
```
